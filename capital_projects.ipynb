## Department of Parks and Recreation Capital Projects

```python
import pandas as pd
import geopandas as gpd
```

```python
df_parks = pd.read_csv('raw_data/Capital_Project_Tracker_20250924.csv')
df_parks
```

```python
df_parks.info()
```

```python
df_parks = df_parks[['DesignStart', 'CurrentPhase', 'Latitude', 'Longitude', 'Title', 'TotalFunding']]
df_parks
```

```python
df_parks = df_parks[df_parks['CurrentPhase'] != 'proposed']
df_parks = df_parks[df_parks['Latitude'] != 0]
```

```python
df_parks
```

```python
df_parks["TotalFunding"] = (df_parks["TotalFunding"]
    .str.replace("Less than $1 million", "$1,000,000")
    .str.replace("Between $1 million and $3 million", "$2,000,000")
    .str.replace("Between $3 million and $5 million", "$4,000,000")
    .str.replace("Between $5 million and $10 million", "$7,500,000")
    .str.replace("Between $5 million and $8 million", "$6,500,000")
    .str.replace("Greater than $10 million", "$10,000,000")
)
```

```python
df_parks['TotalFunding'] = df_parks['TotalFunding'].str.replace('$', '').str.replace(',', '').astype(float)
df_parks = df_parks[df_parks['TotalFunding'] > 0]
df_parks
```

```python
df_parks['DesignStart'] = pd.to_datetime(df_parks["DesignStart"]).dt.year.astype("Int64")
df_parks
```

```python
gdf = gpd.read_file('raw_data/nycd_25c/nycd.shp')
gdf
```

```python
from shapely.geometry import Point

geometry = [Point(xy) for xy in zip(df_parks["Longitude"], df_parks["Latitude"])]
gdf_points = gpd.GeoDataFrame(df_parks, geometry=geometry, crs="EPSG:4326").to_crs("EPSG:2263")
gdf_points
```

```python
gdf_merged = gpd.sjoin(gdf_points, gdf, how="inner", predicate="within")
gdf_merged
```

```python
df = gdf_merged[['DesignStart', 'BoroCD', 'TotalFunding']]
df.sort_values(by='DesignStart')
```

```python
top_5 = df[df["BoroCD"].isin([101, 102, 301, 302, 306])]
bottom_5 = df[df["BoroCD"].isin([202, 205, 311, 312, 317])]
len(top_5), len(bottom_5)
```

```python
# BAN metrics

top_sum = top_5["TotalFunding"].sum()
bottom_sum = bottom_5["TotalFunding"].sum()
difference = top_sum - bottom_sum
mean = gdf_merged['TotalFunding'].agg('mean')
median = gdf_merged['TotalFunding'].agg('median')
difference, mean, median
```

```python
gbt5 = top_5.groupby("DesignStart").agg(
    count=("DesignStart", "count"),           
    total_funding=("TotalFunding", "sum")  
).reset_index()
```

```python
gbb5 = bottom_5.groupby("DesignStart").agg(
    count=("DesignStart", "count"),           
    total_funding=("TotalFunding", "sum")  
).reset_index()

```

```python
gbt5['group'] = 'top 5'
gbb5['group'] = 'bottom 5'
```

```python
gbb5["count"] = gbb5["count"].cumsum()
gbb5["total_funding"] = gbb5["total_funding"].cumsum()

gbt5["count"] = gbt5["count"].cumsum()
gbt5["total_funding"] = gbt5["total_funding"].cumsum()
```

```python
mid_5 = df[df["BoroCD"].isin([414, 405, 408, 307, 316])]

gbm5 = mid_5.groupby("DesignStart").agg(
    count=("DesignStart", "count"),           
    total_funding=("TotalFunding", "sum")  
).reset_index()

gbm5["count"] = gbm5["count"].cumsum()
gbm5["total_funding"] = gbm5["total_funding"].cumsum()

gbm5['group'] = 'middle 5'

gbm5
```

```python
gb = pd.concat([gbb5, gbt5, gbm5])
gb
```

```python
gb.to_csv('cleaned_data/parks-2006-2025.csv', index=False)
```

## NYC Planning Capital Projects

```python
df_top5 = pd.read_csv('raw_data/top_5_capproj.csv')
df_top5.drop(columns = ['the_geom', 'the_geom_webmercator', 'magency', 'maprojid', 'magencyacro', 'sagencyacro'], inplace=True)
df_top5
```

```python
df_bot5 = pd.read_csv('raw_data/bottom_5_capproj.csv')
df_bot5.drop(columns = ['the_geom', 'the_geom_webmercator', 'magency', 'maprojid', 'magencyacro', 'sagencyacro'], inplace=True)
df_bot5
```

```python
# Some projects have no committed funding, this takes their totalspend instead
# These numbers were NOT in the calculation, with them the difference goes up to $6,267,676,214

mask = df_bot5["totalcommit"] == 0
df_bot5["totalcommit"] = df_bot5["totalcommit"].where(~mask, df_bot5["totalspend"])

mask = df_top5["totalcommit"] == 0
df_top5["totalcommit"] = df_top5["totalcommit"].where(~mask, df_top5["totalspend"])
```

```python
top_sum = df_top5['totalcommit'].agg('sum')
bot_sum = df_bot5['totalcommit'].agg('sum')

top_sum - bot_sum
```
